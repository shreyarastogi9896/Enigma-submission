```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
```

```python
profiles=pd.read_excel("train.xlsx")
pairsdata=pd.read_csv("target.csv")
constraints_list = (profiles["Constraints"].dropna().str.split(";").explode().str.strip().str.lower().unique())
interest_list = (profiles["Constraints"].dropna().str.split(";").explode().unique())
print(profiles['Gender'].value_counts())
categorical_cols = ["Gender", "Role", "Industry", "Company_Name", "Location_City", "Seniority_Level"]

for col in categorical_cols:
    profiles[col] = profiles[col].str.strip().str.title()

interest_list
nan_count = profiles.isna().sum()
print(nan_count)
#only small no is empty so trying to find them by same company
#profiles["Company_Size_Employees"] = profiles["Company_Size_Employees"].fillna(profiles["Company_Size_Employees"].median())

profiles["Company_Size_Employees"] = profiles.groupby("Company_Name")["Company_Size_Employees"].transform(lambda x: x.fillna(x.median()))
nan_count = profiles.isna().sum()
print(nan_count)


profiles["Seniority_Level"] = profiles.groupby(["Company_Name", "Role"])["Seniority_Level"].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))
print(profiles["Seniority_Level"].value_counts())
nan_count = profiles.isna().sum()
print(nan_count)


'''profiles['age_group'] = pd.cut(profiles['Age'], bins=range(10, 70, 5), right=False)


for group in profiles['age_group'].unique():
    mask = (profiles['age_group'] == group) & (profiles['Seniority_Level'].isna())
    profiles.loc[mask, 'Seniority_Level'] = profiles.loc[profiles['age_group'] == group, 'Seniority_Level'].mode()[0]'''


profiles['age_group'] = pd.cut(profiles['Age'], bins=range(10, 70, 5), right=False)


for group in profiles['age_group'].unique():
    mask = (profiles['age_group'] == group) & (profiles['Seniority_Level'] == "Unknown")
    mode_value = profiles.loc[(profiles['age_group'] == group) & (profiles['Seniority_Level'] != "Unknown"), 'Seniority_Level'].mode()
    if not mode_value.empty:
        profiles.loc[mask, 'Seniority_Level'] = mode_value[0]



print(profiles["Seniority_Level"].value_counts())

```

```python
df=pairsdata.merge(profiles,left_on="src_user_id",right_on="Profile_ID",how="left",suffixes=("","_src"))
df
df = df.merge(profiles,left_on="dst_user_id",right_on="Profile_ID",how="left",suffixes=("_src", "_dst")
)
df
```

```python
#parsing and converting the multidata columns into numerical ones
def parse(x):
    if pd.isna(x):
        return set()
    return set(item.strip().lower() for item in str(x).split(";") if item.strip())
'''def convert(src_set,dst_set):
    inter=src_set & dst_set
    union=src_set | dst_set
    return {"overlap": len(inter),"jaccard": len(inter) / len(union) if len(union) else 0.0,"src_count": len(src_set),"dst_count": len(dst_set)}'''

cols = ["Business_Interests", "Business_Objectives", "Constraints"]

for col in cols:
    src = df[col + "_src"].apply(parse)
    dst = df[col + "_dst"].apply(parse)

    df[col.lower() + "_overlap"] = [
        len(s & d) for s, d in zip(src, dst)
    ]

    df[col.lower() + "_jaccard"] = [
        len(s & d) / len(s | d) if len(s | d) else 0
        for s, d in zip(src, dst)
    ]

    df[col.lower() + "_src_count"] = [len(s) for s in src]
    df[col.lower() + "_dst_count"] = [len(d) for d in dst]
df
```

```python
df.drop(columns=["Business_Interests_src", "Business_Interests_dst","Business_Objectives_src", "Business_Objectives_dst","Constraints_src", "Constraints_dst"],inplace=True
)
df
```

```python
df["age_diff"] = (df["Age_src"] - df["Age_dst"]).abs()
df["same_gender"] = (df["Gender_src"] == df["Gender_dst"]).astype(int)
df.columns
```

```python
#encoding the required columns
role_encoder = LabelEncoder()

roles = pd.concat([df["Role_src"],df["Role_dst"]])

role_encoder.fit(roles)

df["Role_src_enc"] = role_encoder.transform(df["Role_src"])
df["Role_dst_enc"] = role_encoder.transform(df["Role_dst"])
df = df.drop(columns=["Role_src", "Role_dst"])
```

```python
#encoding industry
industry_encoder = LabelEncoder()

industries = pd.concat([df["Industry_src"],df["Industry_dst"]])

industry_encoder.fit(industries)

df["Industry_src_enc"] = industry_encoder.transform(df["Industry_src"])
df["Industry_dst_enc"] = industry_encoder.transform(df["Industry_dst"])

df = df.drop(columns=["Industry_src", "Industry_dst"])
```

```python
#encoding company
company_encoder = LabelEncoder()

all_companies = pd.concat([df["Company_Name_src"],df["Company_Name_dst"]])

company_encoder.fit(all_companies)

df["Company_Name_src_enc"] = company_encoder.transform(df["Company_Name_src"])
df["Company_Name_dst_enc"] = company_encoder.transform(df["Company_Name_dst"])

df = df.drop(columns=["Company_Name_src", "Company_Name_dst"])
```

```python
#encoding city
city_encoder = LabelEncoder()

cities = pd.concat([df["Location_City_src"],df["Location_City_dst"]])

city_encoder.fit(cities)

df["Location_City_src_enc"] = city_encoder.transform(df["Location_City_src"])
df["Location_City_dst_enc"] = city_encoder.transform(df["Location_City_dst"])

df = df.drop(columns=["Location_City_src","Location_City_dst"])
```

```python
seniority_encoder = LabelEncoder()

seniority = pd.concat([df["Seniority_Level_src"],df["Seniority_Level_dst"]])

seniority_encoder.fit(seniority)

df["Seniority_Level_src_enc"] = seniority_encoder.transform(df["Seniority_Level_src"])
df["Seniority_Level_dst_enc"] = seniority_encoder.transform(df["Seniority_Level_dst"])

df = df.drop(columns=["Seniority_Level_src","Seniority_Level_dst"])

```

```python
# === Add pairwise boolean features ===
df["same_role"] = (df["Role_src_enc"] == df["Role_dst_enc"]).astype(int)
df["same_industry"] = (df["Industry_src_enc"] == df["Industry_dst_enc"]).astype(int)
df["same_company"] = (df["Company_Name_src_enc"] == df["Company_Name_dst_enc"]).astype(int)
df["same_city"] = (df["Location_City_src_enc"] == df["Location_City_dst_enc"]).astype(int)

# === Add differences for numeric features ===
df["seniority_diff"] = abs(df["Seniority_Level_src_enc"] - df["Seniority_Level_dst_enc"])
df["company_size_diff"] = abs(df["Company_Size_Employees_src"] - df["Company_Size_Employees_dst"])

# === Normalized overlaps ===
df["interests_src_ratio"] = df["business_interests_overlap"] / (df["business_interests_src_count"] + 1e-5)
df["interests_dst_ratio"] = df["business_interests_overlap"] / (df["business_interests_dst_count"] + 1e-5)

df["objectives_src_ratio"] = df["business_objectives_overlap"] / (df["business_objectives_src_count"] + 1e-5)
df["objectives_dst_ratio"] = df["business_objectives_overlap"] / (df["business_objectives_dst_count"] + 1e-5)

df["constraints_src_ratio"] = df["constraints_overlap"] / (df["constraints_src_count"] + 1e-5)
df["constraints_dst_ratio"] = df["constraints_overlap"] / (df["constraints_dst_count"] + 1e-5)

# === Interaction features ===
df["interests_jaccard_same_industry"] = df["business_interests_jaccard"] * df["same_industry"]
df["objectives_jaccard_same_role"] = df["business_objectives_jaccard"] * df["same_role"]
df["constraints_jaccard_same_company"] = df["constraints_jaccard"] * df["same_company"]

# === Self-pair flag ===
df["self_pair"] = (df["src_user_id"] == df["dst_user_id"]).astype(int)

#Final X columns
X_columns = [
    'Company_Size_Employees_src','Company_Size_Employees_dst',
    'Age_src','Age_dst','age_diff','same_gender',
    'business_interests_overlap','business_interests_jaccard','business_interests_src_count','business_interests_dst_count',
    'business_objectives_overlap','business_objectives_jaccard','business_objectives_src_count','business_objectives_dst_count',
    'constraints_overlap','constraints_jaccard','constraints_src_count','constraints_dst_count',
    'Role_src_enc','Role_dst_enc','Industry_src_enc','Industry_dst_enc',
    'Company_Name_src_enc','Company_Name_dst_enc','Location_City_src_enc','Location_City_dst_enc',
    'Seniority_Level_src_enc','Seniority_Level_dst_enc',

    'same_role','same_industry','same_company','same_city',
    'seniority_diff','company_size_diff',
    'interests_src_ratio','interests_dst_ratio',
    'objectives_src_ratio','objectives_dst_ratio',
    'constraints_src_ratio','constraints_dst_ratio',
    'interests_jaccard_same_industry','objectives_jaccard_same_role','constraints_jaccard_same_company',
    'self_pair'
]

#Final X and y
x =df[X_columns]
y =df["compatibility_score"]

import xgboost as xgb
model3 = xgb.XGBRegressor(
    n_estimators=500,
    max_depth=8,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    tree_method="hist"
)
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42)
model3.fit(x_train, y_train)

pred_val = model3.predict(x_val)


from sklearn.metrics import mean_squared_error
rmse = mean_squared_error(y_val, pred_val)
print("Validation RMSE:", rmse)
#print("X shape:", X.shape)
#print("Columns in X:", X.columns.tolist())
```
